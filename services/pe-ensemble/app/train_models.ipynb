# Training OPED

```python
from models.oped_wrapper import OPEDModelWrapper
from pe_common.constants import DEVICE, DATA_ROOT, MODEL_ROOT

import pandas as pd

# load the standardized pridict1 library1 dataset
library1_path = DATA_ROOT / 'standardized' / 'pridict1' / 'library1' / 'hek293t-pe2.parquet'
library1_df = pd.read_parquet(library1_path)

# initialize the OPED model
oped_model = OPEDModelWrapper()

# prepare the data for training
train_df, test_df = oped_model.prepare_data(library1_df, split_data=True)

# train the model with a more stable setup
train_hparams = {
    'epoch_num': 100,
    'batch_size': 128,
    'lr': 1e-4,
    'best_epoch': True,
    'grad_clip': 1.0,
    'reshuffle_each_epoch': True,
    'early_stopping_patience': 10,
}
oped_model.train(train_df, val_data=test_df, hyperparameters=train_hparams)

# evaluate the model
oped_model.evaluate(test_df)
```
